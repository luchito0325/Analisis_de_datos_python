```python
# importar librerias
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.dates import DateFormatter
import matplotlib.gridspec as gridspec 
```

```python
month = np.array(['E','F','M','J','A'])
sales = np.array([20,25,30,28,35])
```

```python
# configurar el tamaño del grafico

plt.figure(figsize=(8,6))

# crear el grafico
plt.plot(month ,sales, marker = 'o', color = 'blue')
plt.title('ventas mensuales de un producto')
plt.xlabel('Meses')
plt.ylabel('ventas en miles')

plt.show()
```

```python
# grafico de dispersion

plt.figure(figsize= (8,6))

hours = [1,2,3,4,5,6,7,8]
exam = [55,60,65,65,70,75,80,85]

plt.scatter(hours, exam, color = 'green')

plt.xlabel('Relacion entre horas estuduadas y el puntaje')
plt.ylabel('Horas')
plt.xlabel('Puntaje')

plt.show()

```

```python
# personalizacion de graficos

# grafico de dispersion de dos estudiantes

plt.figure(figsize= (8,6))

hours = [1,2,3,4,5,6,7,8]
exam_score_student_1 = [55,60,65,65,70,75,80,85]
exam_score_student_2 = [55,58,63,69,74,78,83,90]

plt.scatter(hours, exam_score_student_1,marker= 'o', color = 'green', linestyle = '-', label = 'Estudiante 1')
plt.scatter(hours, exam_score_student_2,marker= 's', color = 'red', linestyle = '--', label = 'Estudiante 2')

plt.xlabel('Relacion entre horas estuduadas y el puntaje de dos estuduantes')
plt.ylabel('Horas')
plt.xlabel('Puntaje')

plt.show()
```

```python
# personalizacion de graficos

# grafico de lineas de dos estudiantes

plt.figure(figsize= (8,6))

hours = [1,2,3,4,5,6,7,8]
exam_score_student_1 = [55,60,65,65,70,75,80,85]
exam_score_student_2 = [55,58,63,69,74,78,83,90]

plt.plot(hours, exam_score_student_1,marker= 'o', color = 'green', linestyle = '-', linewidth = 2 , label = 'Estudiante 1')
plt.plot(hours, exam_score_student_2,marker= 's', color = 'red', linestyle = '--', linewidth = 2 , label = 'Estudiante 2')

plt.xlabel('Relacion entre horas estuduadas y el puntaje de dos estuduantes')
plt.ylabel('Puntaje')
plt.xlabel('Horas')

plt.show()
```

```python
# grafico de barras

plt.figure(figsize=(8,6))

categories= ['producto A','producto B','producto C']
sales_prod = [120,150,90]


plt.bar(categories, sales_prod, color = 'skyblue', label = 'ventas mensuales')

# anotacion con flecha para destacar un punto especifico

plt.annotate('Maximo de ventas', xy=('producto B', 150 ),xytext= ('producto C',160),
             arrowprops= dict(facecolor = 'black', shrink = 0.05))

plt.title('ventas de productos en un mes')
plt.xlabel('Productos')
plt.ylabel('Ventas')

#plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8,6)) # crear la figura donde va a estar el grafico

categories= ['producto A','producto B','producto C']
sales_prod = [120,150,90]

plt.barh(categories,sales_prod, color= 'skyblue', label = 'Ventas Mensuales')

plt.title('Ventas de Productos en un mes')
plt.xlabel('Ventas')
plt.ylabel('Productos')

plt.show()
```

```python
#  diagrama de pastel para representar proporciones
# labels = cuales son las categorias de las divisiones del grafico en este caso los productos
# autopct = dar la proporcion a cada una de las partes del diagrama
# startangle  en que angulo tiene que empezar a hacer las divisiones
plt.figure(figsize=(8,6))

products = ['producto A','producto B','producto C']
marker_share = [50,35,15]

plt.pie(marker_share, labels= products, autopct= '%1.1f%%', startangle= 140,
        colors= ['gold','lightcoral','orange'] )

plt.title('Participacion de Mercado por Producto')

#  asegurar que se el grafico sea un circulo

plt.axis('equal')

plt.show()
```

```python
# histogramas para resumir distribucion y dispercion de un conjunto de datos

data = np.random.normal(170,10,200)
print(data)
```

```python
# alpha es la transparencia del grafico 
plt.hist(data, color= 'salmon',bins= 10, edgecolor= 'black', alpha = 0.6)
plt.title('Distribucion de altura')
plt.xlabel('Altura en cms')
plt.ylabel('Densidad')
plt.show()
```

```python
# diagramas de cajas o boxplots resumen la distribucion de datos a travez de cuartiles permite identificar valores atopicos

np.random.seed(0)

# crear una lista de arrays de informacion de edades

ages = [np.random.normal(30,5,100),
        np.random.normal(40,5,100),
        np.random.normal(35,5,100)]

print(ages)
```

```python
# crear diagrama de cajas
# patch_artist = rellena las cajas de color
# notch = añade un recorte para estar concientes del nivel confianza de  los datos
# vert = que el grafico sea vertical

plt.boxplot(ages, patch_artist= True, notch= True, vert= True,
            labels= ['Grupo 1','Grupo 2','Grupo 3'])

plt.title('Distribucion de edades por grupo')
plt.xlabel('Grupo')
plt.ylabel('Edad (años)')

plt.show()
```

```python
#generar un rago de fechas con pandas


dates = pd.date_range(start= '2023-01-03', periods= 100)

values = np.random.rand(100).cumsum()

data = pd.DataFrame({'Date':dates, 'Value':values})

data
```

```python
# grear el grafico de lineas para series temporales

fig, ax = plt.subplots(figsize = (12,6))

ax.plot(data['Date'], data['Value'], color = 'green')

plt.xticks(rotation = 45)
plt.title('Serie de tiempo con formato en la fechas')
plt.xlabel('Date')
plt.ylabel('Value')

plt.show()
```

```python
dates = pd.date_range(start='2023-01-01',periods= 12, freq= 'M')
sales = np.random.randint(1000,5000, size = 12)

sales_data = pd.DataFrame({'Date':dates, 'Sales': sales})

plt.plot(sales_data['Date'], sales_data['Sales'], marker = 'o',
         linestyle = '-', label = 'Ventas Mensuales')

plt.gca().xaxis.set_major_formatter(DateFormatter('%b %Y'))

plt.xticks(rotation = 45)
plt.title('Analisis de ventas mensuales')
plt.xlabel('Date')
plt.ylabel('Sales')

plt.legend()

plt.tight_layout()

plt.show()
print(sales_data.head(12))

```

```python
#crear multiples graficos en una sola figura
# graficar seno

x = np.linspace(0,10,100)
y = np.sin(x)

data_sin = np.random.randn(100)
```

```python
gs = gridspec.GridSpec(2,2,height_ratios=[2,1],width_ratios=[1,1])

fig = plt.figure(figsize=(10,8))

# primer subplot grande ocupara la primera fila
# gs[1,0]  fila 1 columna 0

ax1 = fig.add_subplot(gs[0,:])
ax1.plot(x,y,color = 'blue')
ax1.set_title('Seno de x')
ax1.set_xlabel('x')
ax1.set_ylabel('sin (x)')


# segundo subplot, ocupa la esquina inferior izquierda

ax2 = fig.add_subplot(gs[1,0])
ax2.hist(data_sin,bins= 20, color= 'purple', edgecolor = 'black')
ax2.set_title('Histograma')
ax2.set_xlabel('valor')
ax2.set_ylabel('frecuencia')

# tercer subplot en la esquina inferior derecha
ax3 = fig.add_subplot(gs[1,1])
ax3.scatter(x, y, color = 'red')
ax3.set_title('Dispersion del seno')
ax3.set_xlabel('x')
ax3.set_ylabel('sin (x)')


plt.tight_layout()

plt.show()
```
