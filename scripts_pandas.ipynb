## Manejo de Datos Faltantes en Pandas: Identificación y Tratamiento

```python
# identificar datos faltantes
miising_data = retail_data.isna()
print(miising_data)
```

```python
print(miising_data.head)
```

```python
# validar todas lac columnas con la suma de valores fatantes 
missing_data_count = retail_data.isna().sum()
print('Conteo de datos faltantes por columna: \n', missing_data_count)
```

```python
#  eliminar o sustituir por otra informacion

no_missing_rows = retail_data.dropna()
print("Datos sin filas con valores faltantes: \n", no_missing_rows)
```

```python
# validar todas lac columnas con la suma de valores fatantes 
no_missing_rows_count = no_missing_rows.isna().sum()
print('Conteo de datos faltantes por columna: \n', no_missing_rows_count)
```

```python
no_missing_columns = retail_data.dropna(axis= 1)
print("Datos sin columnas con valores faltantes: \n", no_missing_columns)
```

```python
#  completar todos los valores faltantes con 0 
retail_data_fillef_zeros = retail_data.fillna(0)
print(retail_data_fillef_zeros)
```

```python
# validar todas lac columnas con la suma de valores fatantes 
retail_data_fillef_zeros_conunt = retail_data_fillef_zeros.isna().sum()
print('Conteo de datos faltantes por columna: \n', retail_data_fillef_zeros_conunt)
```

```python
# retail_data_fillef_zeros = retail_data.fillna(0)  ejemplo con los de zeros
mean_unit_price = retail_data['UnitPrice'].mean()
retail_data_filled_mean = retail_data['UnitPrice'].fillna(mean_unit_price)
print(retail_data_filled_mean)
```
