## manipulacion de columas

```python
import pandas as pd
```

```python
df = pd.read_csv('online_retail.csv')
```

```python
# crear una columna a partir de la operacion de dos columnas
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']
print(df.head())
```

```python
# columnas calculadas a partir de condicionales
df['HighValue'] = df['TotalPrice'] > 15
print(df['HighValue'].head(10))
```

```python
df.info()
```

```python
#cambiar el tipo de dato a fecha conversion de tipo

df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
print(df.info())
```

```python
# aplicacion de funciones lambda, este aplica a cada fila seleccionada la operacion que le especifiquemos

df['DiscountedPrice'] = df['UnitPrice'].apply(lambda x:x*0.9)
print(df.head(3))
```

```python
# hacer modificaciones a columnas categoricas
def categorize_price(price):
    if price > 50:
        return 'Hight'
    elif price < 20:
        return 'Medium'
    else:
        'Low'

df['PriceCategory'] = df['UnitPrice'].apply(categorize_price)
print(df.head(3))
```

## Análisis de Ventas con Pandas: Agrupaciones y Estadísticas

```python
# group by

country_count = df['Country'].value_counts()
print(country_count)
```

```python
country_group = df.groupby('Country')['Quantity'].sum()
print(country_group)
```

```python
# estadisticas agrupadas media y suma del preciomunitario por ciudad
country_stats = df.groupby('Country')['UnitPrice'].agg(['mean','sum'])
print(country_stats)
```

```python
# cantidad agrupada por ciudad y po cod de stock
country_stock_group = df.groupby(['Country','StockCode'])['Quantity'].sum()
print(country_stock_group)
```

```python
def total_revenue( group):
    return(group['Quantity']* group['UnitPrice']).sum()

revenue_per_country = df.groupby('Country').apply(total_revenue)
print(revenue_per_country)
```

```python
country_desc = revenue_per_country.sort_values(ascending= False)
#print(country_desc)
print("los paises con mas ventas son: \n" ,country_desc.iloc[:3])

print("los paises con menos ventas son: \n" ,country_desc.iloc[-3:])
```

```python
# CRUCES DE DATAFRAMES MERGE JOIN

import pandas as pd

# Crear DataFrames de ejemplo
df1 = pd.DataFrame({
    'key': ['A', 'B', 'C'],
    'value1': [1, 2, 3]
})

df2 = pd.DataFrame({
    'key': ['B', 'C', 'D'],
    'value2': [4, 5, 6]
})

print(df1)
print(df2)
```

```python
inner_merge = pd.merge(df1,df2, on= 'key', how= 'inner')
inner_merge
```

```python
# combina el total de los datos de los dateframes
outer_merger = pd.merge(df1,df2,on='key', how= 'outer')
outer_merger
```

```python
# merge a la izquierda
left_merge = pd.merge(df1,df2, on= 'key', how= 'left')
left_merge
```

```python
# merge a la izquierda
right_merge = pd.merge(df1,df2, on= 'key', how= 'right')
right_merge
```

```python
# concatenar dataframes vertical u horizontal
# Crear DataFrames de ejemplo
df3 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2']
})
    
df4 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5']
})

print(df3)
print(df4)
```

```python
# concatenar vertical
vertical_concat = pd.concat([df3,df4])

vertical_concat
```

```python
# concatenar horizontal
horizontal_concat = pd.concat([df3,df4],axis= 1)
horizontal_concat

horizontal_concat
```

```python
#  joined unir dateframes por indices o columna clave

# Crear DataFrames de ejemplo con índices
df5 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2']
}, index=['K0', 'K1', 'K2'])

df6 = pd.DataFrame({
    'C': ['C0', 'C1', 'C2'],
    'D': ['D0', 'D1', 'D2']
}, index=['K0', 'K2', 'K3'])

print(df5)
print(df6)
```

```python
joined = df5.join(df6,how='inner')
joined
```
