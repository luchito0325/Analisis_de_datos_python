## manipulacion de columas

```python
import pandas as pd
```

```python
df = pd.read_csv('online_retail.csv')
```

```python
# crear una columna a partir de la operacion de dos columnas
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']
print(df.head())
```

```python
# columnas calculadas a partir de condicionales
df['HighValue'] = df['TotalPrice'] > 15
print(df['HighValue'].head(10))
```

```python
df.info()
```

```python
#cambiar el tipo de dato a fecha conversion de tipo

df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
print(df.info())
```

```python
# aplicacion de funciones lambda, este aplica a cada fila seleccionada la operacion que le especifiquemos

df['DiscountedPrice'] = df['UnitPrice'].apply(lambda x:x*0.9)
print(df.head(3))
```

```python
# hacer modificaciones a columnas categoricas
def categorize_price(price):
    if price > 50:
        return 'Hight'
    elif price < 20:
        return 'Medium'
    else:
        'Low'

df['PriceCategory'] = df['UnitPrice'].apply(categorize_price)
print(df.head(3))
```

## Análisis de Ventas con Pandas: Agrupaciones y Estadísticas

```python
# group by

country_count = df['Country'].value_counts()
print(country_count)
```

```python
country_group = df.groupby('Country')['Quantity'].sum()
print(country_group)
```

```python
# estadisticas agrupadas media y suma del preciomunitario por ciudad
country_stats = df.groupby('Country')['UnitPrice'].agg(['mean','sum'])
print(country_stats)
```

```python
# cantidad agrupada por ciudad y po cod de stock
country_stock_group = df.groupby(['Country','StockCode'])['Quantity'].sum()
print(country_stock_group)
```

```python
def total_revenue( group):
    return(group['Quantity']* group['UnitPrice']).sum()

revenue_per_country = df.groupby('Country').apply(total_revenue)
print(revenue_per_country)
```

```python
country_desc = revenue_per_country.sort_values(ascending= False)
#print(country_desc)
print("los paises con mas ventas son: \n" ,country_desc.iloc[:3])

print("los paises con menos ventas son: \n" ,country_desc.iloc[-3:])
```
